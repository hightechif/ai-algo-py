# Logistic Regression: Breast Cancer Classification
Let's test our pure NumPy `LogisticRegressionFromScratch` and `sklearn.linear_model.LogisticRegression` on the Breast Cancer binary classification dataset.

```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

from scratch import LogisticRegressionFromScratch
from framework import fit_and_predict

# 1. Load Breast Cancer data
dataset = datasets.load_breast_cancer()
X, y = dataset.data, dataset.target

# 2. Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1234)
```

```python
# 3. Run From Scratch Model
# Note: Breast cancer features are unscaled by default, so we lower the learning rate
scratch_model = LogisticRegressionFromScratch(learning_rate=0.0001, n_iters=1000)
scratch_model.fit(X_train, y_train)
predictions_scratch = scratch_model.predict(X_test)

def accuracy(y_true, y_pred):
    accuracy = np.sum(y_true == y_pred) / len(y_true)
    return accuracy

scratch_acc = accuracy(y_test, predictions_scratch)
print("Scratch Accuracy:", scratch_acc)
```

```python
# 4. Run scikit-learn Framework Model
framework_model, predictions_framework, framework_acc = fit_and_predict(X_train, y_train, X_test, y_test)
print("Framework Accuracy:", framework_acc)
```

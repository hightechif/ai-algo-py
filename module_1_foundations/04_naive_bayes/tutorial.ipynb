# Tutorial: Multinomial Naive Bayes using 20 Newsgroups Classification
In this notebook, we evaluate our from-scratch `scratch.py` implementation of Multinomial Naive Bayes vs Scikit-Learn utilizing a real-world text classification dataset.

```python
import numpy as np
from scratch import MultinomialNaiveBayes
from framework import ScikitLearnMultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split

# Load the 20 newsgroups dataset for binary classification
categories = ['alt.atheism', 'comp.graphics']
train_data = fetch_20newsgroups(subset='train', categories=categories, remove=('headers', 'footers', 'quotes'))
test_data = fetch_20newsgroups(subset='test', categories=categories, remove=('headers', 'footers', 'quotes'))

X_train_text, y_train = train_data.data, train_data.target
X_test_text, y_test = test_data.data, test_data.target

print(f"Training samples: {len(X_train_text)}")
print(f"Testing samples:  {len(X_test_text)}")
```

## Generating the Feature Vector
We utilize `CountVectorizer` to generate a term-frequency matrix (Bag of Words) mapping the strings to integer token counts. This will result in a much larger vocabulary than our initial mock dataset.

```python
vectorizer = CountVectorizer(stop_words='english', max_features=5000)
X_train = vectorizer.fit_transform(X_train_text).toarray()
X_test = vectorizer.transform(X_test_text).toarray()

print(f"Vocabulary size: {X_train.shape[1]}")
print(f"Vectorized shape: {X_train.shape}")
```

## Comparing Models
Let's train both the from-scratch version and scikit-learn. With a larger dataset, we expect both models to perform significantly better and maintain consistent accuracy scores.

```python
# Scratch
nb_scratch = MultinomialNaiveBayes(alpha=1.0)
nb_scratch.fit(X_train, y_train)

# Scikit-Learn
nb_sklearn = ScikitLearnMultinomialNB()
nb_sklearn.fit(X_train, y_train)

print(f"Scratch Accuracy: {nb_scratch.score(X_test, y_test):.4f}")
print(f"Sklearn Accuracy: {nb_sklearn.score(X_test, y_test):.4f}")
```
